```python
from rich.console import Console
import scipy.stats as stats
import numpy as np
import matplotlib.pyplot as polt

console = Console()
```

```python
alpha = 0.05

n = 3
m = 2

x_n = np.array([-1.11, -6.10, 2.42])
y_m = np.array([-2.29, -2.91])

sigma_x = 2
sigma_y = 1

delta = x_n.mean() - y_m.mean()

# Квантиль
N = stats.norm(loc=0, scale=np.sqrt(sigma_x / n + sigma_y / m))
q = N.ppf(1 - alpha)
p_value = N.sf(delta)
```

Построить график мощности критерия

```python
_, axs = polt.subplots(1, 1, figsize=(10, 6))

axs.set_title("Мощность критерия")
axs.set_xlabel("d")
axs.set_ylabel("W")

d = np.linspace(0, 5, 100)

axs.plot(d, N.sf(-d + q), label="Мощность критерия", color="red")

polt.grid(which='minor', linestyle='--', linewidth=1)
polt.grid(which='major', linewidth=1)

polt.show()
```
